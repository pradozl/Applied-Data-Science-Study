```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Carregar os dados
df = pd.read_csv('dataset_ESAVI_limpo.csv')
```

```python
# 2. Seleção de Variáveis para PCA
# O PCA exige apenas dados numéricos, por isso é preciso tratar as categorias.
# Peguei 'Idade Evento' e transformei outras importantes em binárias (Dummies)
cols_interesse = ['Idade Evento', 'Sexo', 'Raça/Cor', 'Houve atendimento médico?', 'Tipo de Evento']
df_pca = df[cols_interesse].copy()
```

```python
# 3. Tratamento de Dados Categóricos (One-Hot Encoding)
# Transformei colunas de texto em colunas numéricas (0 ou 1)
df_pca_prepared = pd.get_dummies(df_pca, columns=['Sexo', 'Raça/Cor', 'Houve atendimento médico?', 'Tipo de Evento'], drop_first=True)
```

```python
# 4. Limpeza de valores nulos (Requisito do PCA)
df_pca_prepared = df_pca_prepared.dropna()

print(f"Dataset preparado para PCA: {df_pca_prepared.shape[1]} dimensões (características).")
```

```python
# 5. PADRONIZAÇÃO (SCALING)

# Criei o objeto scaler
scaler = StandardScaler()

# Ajustei e transformei os dados
# Isso coloca todos os dados na mesma escala (Média 0, Desvio Padrão 1)
dados_padronizados = scaler.fit_transform(df_pca_prepared)

# Convertendo de volta para DataFrame para visualização
df_final = pd.DataFrame(dados_padronizados, columns=df_pca_prepared.columns)

print("\n--- Dados Padronizados (Primeiras 5 linhas) ---")
print(df_final.head())
```

### Interpretação da Preparação:
1. **Dummies**: Como o PCA é uma técnica geométrica baseada em variância, transformamos categorias em eixos numéricos.
2. **StandardScaler**: Esta é a etapa mais crucial. Sem o scaling, se a 'Idade' variasse de 0 a 90 e uma 
   variável binária variasse de 0 a 1, o PCA acharia que a Idade é 90 vezes mais importante apenas pela escala. 
   Agora, todas as variáveis 'competem' em igualdade no espaço n-dimensional.
3. **Próximo Passo**: Os 'dados_padronizados' agora podem ser alimentados no algoritmo PCA do sklearn.
