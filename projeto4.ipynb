```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('dataset_ESAVI_limpo.csv')
```

```python
df['Data da Notificação'] = pd.to_datetime(df['Data da Notificação'])
```

```python
# --- GRÁFICO 1: HISTOGRAMA (Idade) ---
plt.figure(figsize=(10, 6))
sns.histplot(df['Idade Evento'], bins=20, kde=True, color='teal')
plt.title('Distribuição Etária das Notificações de ESAVI')
plt.xlabel('Idade do Paciente')
plt.ylabel('Quantidade de Casos')
plt.show()
```

```python
# --- GRÁFICO 2: BOXPLOT (Idade vs Gravidade) ---
plt.figure(figsize=(10, 6))
sns.boxplot(x='Classificação de gravidade', y='Idade Evento', data=df, palette='OrRd')
plt.title('Dispersão de Idade por Gravidade do Evento')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# --- GRÁFICO 3: SCATTER PLOT (Tempo vs Idade) ---
plt.figure(figsize=(12, 6))
sns.scatterplot(x='Data da Notificação', y='Idade Evento', hue='Tipo de Evento', data=df, alpha=0.6)
plt.title('Evolução das Notificações no Tempo por Idade')
plt.xticks(rotation=45)
plt.show()
```

```python
# A Análise Exploratória (EDA) realizada através da biblioteca Matplotlib/Seaborn permitiu extrair padrões ocultos no dataset ESAVI_Goiás
# Através da técnica de Histograma com KDE, confirmámos que a distribuição de eventos não segue uma curva normal, concentrando-se fortemente na faixa etária pediátrica (3-12 anos).
# A visualização de Dispersão (Scatter Plot) revelou que os 'Erros de Imunização' não são eventos isolados cronologicamente, sugerindo que o sistema de vigilância é sensível o
# suficiente para captar falhas processuais contínuas. Além disso, a aplicação de Boxplots por Categoria de Gravidade demonstrou que a idade não é o único fator determinante
# para a gravidade do evento, dado que a mediana de idade permaneceu estável entre casos graves e não graves.
```
